```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import convolve
```

```python
input_path = "image.png"
img = cv2.imread(input_path,cv2.IMREAD_GRAYSCALE)
plt.figure(figsize=(6,4))
plt.imshow(img,cmap='gray')
#plt.show()
```

```python
sharpen=[[0,-1,0],  #to focus more on brighter parts of the  image 
        [-1,5,-1],
        [0,-1,0]]
```

```python
kernel=sharpen
#performing the convolution operation
im=np.array(img,dtype=float)   #by default  it is in integer part 
img2=convolve(im,kernel)
plt.figure(figsize=(6,4))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2,cmap='gray')
```
